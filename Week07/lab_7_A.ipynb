## Exercise 1: Linked Lists

A **linked list** is a data structure for storing a sequence of items.  Each item in the list is called a **node**.  Unlike a regular Python list, the items do not need to be stored next to each other in memory - instead, each node holds a **reference** (sometimes called a *pointer* or *link*) to the next node in the sequence.

Each node contains two things:
* The **data** it is storing (for example, a price or a name)
* A reference to the **next** node - or `None` if this is the last node in the list

In Python, we represent a node using a class.  Here is a simple node class for an invoice system, where each node stores the price of one item on the invoice:

> ```python
> class InvoiceItem:
>     def __init__(self, price):
>         self.price = price
>         self.next = None
> ```

We can create an instance of this class just like any other class:

> ```python
> item1 = InvoiceItem(12.99)
> print("Price:", item1.price)
> ```

Output:

> ```text
> Price: 12.99
> ```

### Exercise 1.10

So far we have been managing the linked list with standalone functions and separate `head` and `tail` variables.  A cleaner approach is to **encapsulate** everything into a class.  The class keeps track of `head` and `tail` internally, and provides methods for all the common operations.

Here is a `LinkedList` class for the invoice example, bringing together everything from the previous exercises:

> ```python
> class LinkedList:
>     def __init__(self):
>         self._head = None
>         self._tail = None
>     
>     def append(self, price):
>         new_item = InvoiceItem(price)
>         if self._head is None:
>             self._head = new_item
>             self._tail = new_item
>         else:
>             self._tail.next = new_item
>             self._tail = new_item
>     
>     def get(self, n):
>         current = self._head
>         for i in range(n):
>             current = current.next
>         return current
>     
>     def insert_after(self, n, price):
>         node = self.get(n)
>         new_item = InvoiceItem(price)
>         new_item.next = node.next
>         node.next = new_item
>     
>     def delete_after(self, n):
>         node = self.get(n)
>         node.next = node.next.next
>     
>     def print_all(self):
>         current = self._head
>         while current is not None:
>             print(current.price)
>             current = current.next
> ```

You are not expected to be able to write something like this yourself.  It is shown here to demonstrate how everything we have built across these exercises can be packaged up neatly into a reusable object.  However, you should be able to *use* a class like this.

A `ChatHistory` class, equivalent to the `LinkedList` above, has been written for you in the code cell below.  **Use the `ChatHistory` class to build a short conversation, insert a new message, and print the result.**

````python
# This ChatMessage class is provided for this exercise
class ChatMessage:
    def __init__(self, message):
        self.message = message
        self.next = None

# This ChatHistory class is provided for you - you do not need to write it yourself
class ChatHistory:
    def __init__(self):
        self._head = None
        self._tail = None

    def append(self, message):
        new_node = ChatMessage(message)
        if self._head is None:
            self._head = new_node
            self._tail = new_node
        else:
            self._tail.next = new_node
            self._tail = new_node

    def get(self, n):
        current = self._head
        for i in range(n):
            current = current.next
        return current

    def insert_after(self, n, message):
        node = self.get(n)
        new_node = ChatMessage(message)
        new_node.next = node.next
        node.next = new_node

    def delete_after(self, n):
        node = self.get(n)
        node.next = node.next.next

    def print_all(self):
        current = self._head
        while current is not None:
            print(current.message)
            current = current.next


# Use the ChatHistory class below - write your code here
````
