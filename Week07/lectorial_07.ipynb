## Exercise 2: Linked Lists

<p style="text-align: center; font-style: italic"> We have a <b>queue of customers</b> at a Bank.  Customers join the <b>end of the queue</b>, in order. Bank Cashiers always see whoever is currently at the <b>front of the queue</b></p>

* Implement the above scenario you will need:
    * A class representing your list
    * A class representing an element in the linked list (A customer)
    * Two class variables holding the first and last element of the list
    * A method to add a customer to the end of the list.
    * A method to get the next customer from the start of the list.
    

Hint: you don’t need to arbitrarily add/remove elements from your list, this should simplify your code.
