### Exercise 2.3

```python
time_periods = [ 60, 60, 24, 365 ]
total = 1

for i in range(len(time_periods)):
    total *= time_periods[i]


print("Number of seconds in a year:")
print(total)
```

### Exercise 2.4

```python
time_periods = [ 60, 60, 24, 365 ]
total = 1

for time_period in time_periods:
    total *= time_period


print("Number of seconds in a year:")
print(total)
```

### Exercise 2.5

```python
prices = [10.00, 5.50, 3.20]
tax_rate = 0.2

for price in prices:
    print(price * (1.0 + tax_rate))
```

### Exercise 2.6

```python
prices = [10.00, 5.50, 3.20]
tax_rate = 0.2

for i in range(len(prices)):
    prices[i] = prices[i] * (1.0 + tax_rate)


# The line below prints the prices, which should now include tax
print(prices)
```
