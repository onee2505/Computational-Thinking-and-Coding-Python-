### Exercise 1 - Understanding Scope

Let's try to understand how scope works with some more examples. Below, I have shown 4 examples of Python code. Your task is to predict what the code will print in each case after carefully inspecting the scope of each variable. Then, once you have written your prediction down, run the code and see what gets printed. Does it match what you had in mind? If not, try looking at the code again to understand why.

#### Exercise 1.1

````python
a = 2

def test_scope():
    a = 5

test_scope()
print(a)
````

---
#### Exercise 1.2

````python
a = 2

def test_scope(a):
    a = 5
    return a

test_scope(a)
print(a)
````

Was that the output you expected? But we did return a! Look at it again!

---

#### Exercise 1.3

````python
a = 2

def test_scope(a):
    a = 5
    return a

a = test_scope(a)
print(a)
````

---
#### Exercise 1.4

````python
a = 2

def test_scope():
    a = a + 5
    return a

a = test_scope()
print(a)
````

Did you get an **UnboundLocalError**? This is because you tried to modify a global variable inside of a function.   

Notice the difference between the code in 1.3 and 1.4? In 1.3, we passed the value of 'a' to the function which created a local copy of a, edited it, and returned it. In 1.4, we did not define the function with a parameter. This meant the function tried to modify the value of the global 'a'
