## Exercise 2: Iteration and lists

Loops and lists are frequently used together.  It is very common for programs to achieve a task by looping over each element of a list.  Consider this program which prints every element in a list:

> ```python
> temperatures = [ 23, 24, 24, 22, 19 ]
> 
> for i in range(5):
>     print(temperatures[i])
> ```

The `for` loop will run 5 times, and each time will increment the loop counter `i`.  The loop counter will go from 0 up to 4.  Notice that this sequence starting from zero is also consistent with list indexes starting from zero.

Each time it loops, it will print `temperatures[i]`.  This is the list indexing notation we used above, except using a _variable_ as the index, instead of a fixed number.  The first time the loop runs, `i` is `0`, so it will print `temperatures[0]`.  The next time, `i` will be `1`, and it will print `temperatures[1]`, and so on.

### Exercise 2.4

So far, we have used loops which keep track of the _list index_, and used that to look up elements in the list.  However, there is actually another, simpler way to use `for` loops with lists.  Technically, the structure of a `for` loop is this:

> ```python
> for <variable> in <iterable>:
>     ...
> ```

An _iterable_ is something which can be iterated over, i.e. some ordered sequence of values.  So far, we have used the iterables which are generated by `range(n)`, which gives a sequence of numbers from 0 to _n - 1_.  But actually, lists themselves are iterable.  This means we can do this:

> ```python
> weekdays = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]
> 
> for day in weekdays:
>     print(day)
> ```

Notice there is no use of `range` here, and that this time, the `for` loop variable `day` does not store the index - instead, it is the actual string value from the list itself.  So, the first time the loop runs, `day` will have the value `"Monday"`.  The next iteration, it will have the value `"Tuesday"`, and so on.

This is much nicer to read, and also avoids us having to use the list indexing to look up elements of the list.
