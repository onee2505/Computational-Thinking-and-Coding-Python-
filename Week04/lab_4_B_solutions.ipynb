### Exercise 1.1

### Exercise 1.3

```python
numbers = [ 4, 8, 3, 6, 1, 9, 0 ]

for number in numbers:
    if number % 2 == 0:
        print("Even")
    else:
        print("Odd")
```

### Exercise 1.4

```python
ages = [ 35, 70, 18, 5, 17, 65, 20 ]

for age in ages:
    if age < 18:
        print("Child")
    elif age >= 65:
        print("Senior")
    else:
        print("Adult")
```

### Exercise 1.5

```python
letters = [ "c", "a", "a", "b", "c", "b", "a", "c", "a" ]

a_count = 0
b_count = 0
c_count = 0

for letter in letters:
    if letter == "a":
        a_count += 1
    elif letter == "b":
        b_count += 1
    elif letter == "c":
        c_count += 1

print(a_count)
print(b_count)
print(c_count)
```

### Exercise 1.6

```python
numbers = [ 10, 4, -2, 1, -5, 0, 6 ]

smallest = 99999999

for number in numbers:
    if number < smallest:
        smallest = number

print(smallest)
```

### Exercise 1.7

```python
directions = [ "up", "up", "down", "up", "down", "down", "down", "stop", "up", "up" ]

i = 0
counter = 0

while (directions[i] != "stop"):
    if directions[i] == "up":
        counter += 1
    elif directions[i] == "down":
        counter -= 1
    
    i += 1

print(counter)
```
