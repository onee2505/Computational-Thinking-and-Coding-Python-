### Exercise 1.3

One of the arithmetic operators you have already seen in Python is `%`, which is the [_modulo operator_](https://en.wikipedia.org/wiki/Modulo).  This operator gives the _remainder after division_.  For example, `5 % 2` is `1`, because when you divide 5 by 2 you are left with a remainder of 1.

A common use of this operator is determining whether a number is even or odd:
* If `x % 2` is equal to zero, then `x` is even
* If `x % 2` is equal to one, then `x` is odd

**Write a program which iterates through a list, printing "Even" for the even numbers and "Odd" for the odd numbers.**

At the start of your program, create a list containing these numbers:

`4, 8, 3, 6, 1, 9, 0`

Your program should print:

```text
Even
Even
Odd
Even
Odd
Odd
Even
```

### Exercise 1.7

**Write a program which processes a list of strings which are either "up" or "down", and either increase or decrease a counter accordingly, but stop processing when the list contains "stop".**

At the start of your program, create a list containing these strings:

`"up", "up", "down", "up", "down", "down", "down", "stop", "up", "up"`

The program should maintain a counter, which starts at zero.  Every time there is an "up" in the list, you should add 1 to the number, and every time there is "down", you subtract 1.  However, **if you encounter "stop" in the list, stop counting at that point**.

For the list above, there are 3 ups and 4 downs before the "stop" command, so it should print the result `-1`.

_Hint: the goal of this exercise is to get you to think carefully about which type of loop to use, and how the condition controlling the loop should be defined._
