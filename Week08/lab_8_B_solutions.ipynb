### Exercise 1.6

```python
class AnimalNode:
    def __init__(self, name):
        self.name = name
        self.children = []

# Animal taxonomy tree set up here
animals  = AnimalNode('Animals')
mammals  = AnimalNode('Mammals')
reptiles = AnimalNode('Reptiles')
dog      = AnimalNode('Dog')
cat      = AnimalNode('Cat')
snake    = AnimalNode('Snake')
animals.children  = [mammals, reptiles]
mammals.children  = [dog, cat]
reptiles.children = [snake]

def find_animal(node, name):
    if node.name == name:
        return node
    for child in node.children:
        result = find_animal(child, name)
        if result is not None:
            return result
    return None

print('Found Cat:', find_animal(animals, 'Cat') is not None)
print('Found Dragon:', find_animal(animals, 'Dragon') is not None)
```

### Exercise 1.7

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
        self.children = []

# This tree is set up for you - do not change this code
ceo         = Employee('CEO',         80000)
cto         = Employee('CTO',         60000)
cfo         = Employee('CFO',         60000)
lead_dev    = Employee('Lead Dev',    45000)
lead_design = Employee('Lead Design', 45000)
accountant  = Employee('Accountant',  40000)
ceo.children = [cto, cfo]
cto.children = [lead_dev, lead_design]
cfo.children = [accountant]

def highest_paid(node):
    best = node
    for child in node.children:
        candidate = highest_paid(child)
        if candidate.salary > best.salary:
            best = candidate
    return best

# Do not change this code
best = highest_paid(ceo)
print('Highest paid:', best.name, '(' + str(best.salary) + ')')
```

### Exercise 1.8

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
        self.children = []

# This tree is set up for you - do not change this code
ceo         = Employee('CEO',         80000)
cto         = Employee('CTO',         60000)
cfo         = Employee('CFO',         60000)
lead_dev    = Employee('Lead Dev',    45000)
lead_design = Employee('Lead Design', 45000)
accountant  = Employee('Accountant',  40000)
ceo.children = [cto, cfo]
cto.children = [lead_dev, lead_design]
cfo.children = [accountant]

def count_above(node, threshold):
    count = 1 if node.salary > threshold else 0
    for child in node.children:
        count += count_above(child, threshold)
    return count

# Do not change this code
print('Earning over 50000:', count_above(ceo, 50000))
```

### Exercise 1.9

```python
class MatchNode:
    def __init__(self, name):
        self.name = name
        self.children = []

# This tree is set up for you - do not change this code
final   = MatchNode('Final')
semi1   = MatchNode('Semi 1')
semi2   = MatchNode('Semi 2')
united  = MatchNode('United')
city    = MatchNode('City')
arsenal = MatchNode('Arsenal')
chelsea = MatchNode('Chelsea')
final.children = [semi1, semi2]
semi1.children = [united, city]
semi2.children = [arsenal, chelsea]

def print_at_depth(node, target_depth, current_depth):
    if current_depth == target_depth:
        print(node.name)
    else:
        for child in node.children:
            print_at_depth(child, target_depth, current_depth + 1)

print('Teams at depth 2:')
print_at_depth(final, 2, 0)
```

### Exercise 1.10

```python
class ProductNode:
    def __init__(self, name, price=0):
        self.name = name
        self.price = price
        self.children = []

catalogue   = ProductNode('Catalogue')
electronics = ProductNode('Electronics')
clothing    = ProductNode('Clothing')
phone       = ProductNode('Phone',   499)
laptop      = ProductNode('Laptop',  999)
shirt       = ProductNode('Shirt',    25)
jeans       = ProductNode('Jeans',    60)
catalogue.children   = [electronics, clothing]
electronics.children = [phone, laptop]
clothing.children    = [shirt, jeans]

def print_priced(node):
    if node.price > 0:
        print(node.name + ' - £' + str(node.price))
    for child in node.children:
        print_priced(child)

print_priced(catalogue)
```

```python
class GeoNode:
    def __init__(self, name):
        self.name = name
        self.children = []

# This tree is set up for you - do not change this code
world   = GeoNode('World')
europe  = GeoNode('Europe')
asia    = GeoNode('Asia')
france  = GeoNode('France')
germany = GeoNode('Germany')
japan   = GeoNode('Japan')
china   = GeoNode('China')
world.children  = [europe, asia]
europe.children = [france, germany]
asia.children   = [japan, china]

def print_with_level(node, depth):
    print('Level ' + str(depth) + ': ' + node.name)
    for child in node.children:
        print_with_level(child, depth + 1)

print_with_level(world, 0)
```

### Exercise 1.13

`show_descendants` uses post-order traversal: process all children first, then print the current node's count.

```python
class MatchNode:
    def __init__(self, name):
        self.name = name
        self.children = []

# This tree is set up for you - do not change this code
final   = MatchNode('Final')
semi1   = MatchNode('Semi 1')
semi2   = MatchNode('Semi 2')
united  = MatchNode('United')
city    = MatchNode('City')
arsenal = MatchNode('Arsenal')
chelsea = MatchNode('Chelsea')
final.children = [semi1, semi2]
semi1.children = [united, city]
semi2.children = [arsenal, chelsea]

def show_descendants(node):
    count = 0
    for child in node.children:
        count += 1 + show_descendants(child)
    print(node.name + ': ' + str(count) + ' descendants')
    return count

show_descendants(final)
```
