### Exercise 1.5

```python
class JobRole:
    def __init__(self, role_name, budget, expenses):
        self.role_name = role_name
        self.children = []
        self.budget = budget
        self.expenses = expenses


# The tree setup code below is provided for you - do not change it
ceo = JobRole("CEO", 10000, 850)
cto = JobRole("CTO", 3000, 540)
cfo = JobRole("CFO", 3000, 230)
lead_dev = JobRole("Lead Developer", 100, 50)
lead_design = JobRole("Lead Designer", 100, 100)
accountant = JobRole("Accountant", 150, 200)
analyst = JobRole("Analyst", 150, 130)

ceo.children = [cto, cfo]
cto.children = [lead_dev, lead_design]
cfo.children = [accountant, analyst]

print_tree(ceo, "role_name", "budget", "expenses")
```

### Exercise 1.6

```python
def print_roles(node):
    print(node.role_name)
    for child in node.children:
        print_roles(child)


print_roles(ceo)
```

### Exercise 1.7

```python
def allocate_budget(node, budget):
    print(f"{node.role_name} receives budget: {budget}")
    node.budget = budget

    if len(node.children) > 0:    
        split = budget / len(node.children)

        for child in node.children:
            allocate_budget(child, split)


allocate_budget(ceo, 10000)
```

### Exercise 1.8

```python
def print_roles_post(node):
    for child in node.children:
        print_roles_post(child)
    print(node.role_name)


print_roles_post(ceo)
```

### Exercise 1.9

```python
def total_expenses(node):
    if len(node.children) == 0:
        print(f"{node.role_name} expense = {node.expenses}")
        return node.expenses
    else:
        total = 0
        for child in node.children:
            total += total_expenses(child)

        node.expenses = total
        print(f"{node.role_name} total expense = {total}")
        return node.expenses


total_expenses(ceo)
```
