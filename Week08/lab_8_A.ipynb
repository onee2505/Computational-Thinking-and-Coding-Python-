## Exercise 1: Trees

Trees are a data structure where each item (called a **node**) can be linked to multiple other nodes, called its **children**. This is different from a linked list, where each node has exactly one `next` pointer. While a linked list is a linear structure, a tree is a **hierarchical** structure.

For example, consider this tree which is used to represent **product categories in a retail application**:

> ```text
> Clothing
> ├── Casual
> │   ├── Shirts
> │   └── Trousers
> └── Nightwear
>     └── Pyjamas
> ```

The node at the top of the tree is called the **root**. Nodes which have no children are called **leaves**.

In Python, we can represent a tree node using a class. For example, here is a tree node representing a product category:

> ```python
> class ProductCategory:
>     def __init__(self, name):
>         self.name = name
>         self.children = []
> ```

* `name` stores the name of the product category
* `children` is a list which will hold the child nodes - it starts as an empty list

### Exercise 1.7

A **pre-order traversal** is useful when a parent node must be processed **before** its children. 

Consider our job hierarchy example.  Let's say that in this workplace, budgets are allocated at the top of the hierarchy (i.e. to the CEO), and are then split equally down the hierarchy.  This means each node must have its budget allocated _before_ its children.  Therefore, **pre-order** traversal makes the most sense to do this.

Below is an example of a function which uses pre-order traversal to allocate budgets.  You are not expected to write a function like this yourselves, but see if you can spot the fact that the budgets are allocated _before_ the code loops over the children.  This makes it pre-order.

_Note: This exercise continues from Exercise 1.6 - make sure you have run that cell first._

**Complete the code below to call the `allocate_budget` function**, starting from `ceo`, with a total budget of `10000`.

````python
def allocate_budget(node, budget):
    print(f"{node.role_name} receives budget: {budget}")
    node.budget = budget

    if len(node.children) > 0:    
        split = budget / len(node.children)

        for child in node.children:
            allocate_budget(child, split)


# Call allocate_budget below, starting from ceo with a budget of 10000

````

### Exercise 1.9

A **post-order traversal** is useful when a parent's result **depends on values from all of its children**.

Consider our job role hierarchy.  Let's say that in this workplace, employee expenses travel up the hierarchy - low-level employees record their expenses, and their manager's expenses are the sum of all their children, up to the CEO who has the sum of all the expenses in the business.  As each node's expenses depends on it's children's expenses to be calculated first this suits **post-order traversal**.

Below is an example of a function which uses pre-order traversal to total up expenses.  You are not expected to write a function like this yourselves, but see if you can spot the fact that the budgets are allocated _after_ the code loops over the children.  This makes it post-order.

_Note: This exercise continues from Exercise 1.6 - make sure you have run that cell first._

**Complete the code below to call the `total_expenses` function**, starting from `ceo`.
```

````python
def total_expenses(node):
    if len(node.children) == 0:
        print(f"{node.role_name} expense = {node.expenses}")
        return node.expenses
    else:
        total = 0
        for child in node.children:
            total += total_expenses(child)

        node.expenses = total
        print(f"{node.role_name} total expense = {total}")
        return node.expenses


# Call total_expenses below, starting from ceo.

````
