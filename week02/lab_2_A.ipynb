## Exercise 1: `while` loops

Computer programs often need to _repeat_ a piece of code many times.  The proper term for this is _iteration_.  We can achieve this using loops.

A simple loop provided by Python is the `while` loop.  The loop requires a true or false _condition_ (just like an `if` statement), and it will repeat the code block below it as long as the condition is true.  As soon as the condition is false, it will move on.

> ```python
> while <condition>:
>    <thing to repeat>
> ```

For example, here is a program which counts to 10:

> ```python
> counter = 1
> 
> while counter <= 10:
>     print(counter)
>     counter = counter + 1
> 
> print("Finished counting.")
> ```

When we run this program, here is what happens:
* When the while loop is first executed, `counter` is `1`
* Therefore, the condition `counter <= 10` is true, and Python executes the code block under the loop.
* `1` is printed
* `counter` is _incremented_ (increased by 1)
* The condition `counter <= 10` is checked again - it is still true, so Python executes the block again.
* `2` is printed
* `counter` is _incremented_ (increased by 1)
* _...This continues until the counter reaches `10`..._
* The condition `counter <= 10` is checked again - it is still true, so Python executes the block again.
* `10` is printed
* `counter` is _incremented_ (increased by 1)
* The condition `counter <= 10` is checked again - it is now false, since counter is `11`
* The while loop ends, and we move on to the final `print`.

## Exercise 2: `for` loops

For Exercises 1.1, 1.2 and 1.3, the purpose of the loop was run for a fixed number of iterations.  We used a counter, which we incremented or decremented during each loop.  Because this is a very common pattern, there is a simpler way to write a loop like this, using a `for` loop instead.  Consider this program, counting from 0 to 9:

> ```python
> counter = 0
> 
> while (counter < 10):
>     print(counter)
>     counter += 1
> ```

This can be written as a `for` loop like this:

> ```python
> for counter in range(10):
>     print(counter)
> ```

The line `for counter in range(10)` means: "Create a variable called `counter`, and loop for each of the values in `range(10)`, settings `counter` as the current value each time".  The `range` function generates a sequence of numbers from 0 up to 9.  So, overall, the line is telling Python to loop 10 times.  This is exactly the same end result as the `while` version above.

_Note:  Previously we wrote loops counting from 1 to 10, but this time we are counting from 0 to 9.  Because zero is included, **this is still ten iterations** - even though we stop at 9!  Programming languages almost always count from zero, not one, which you will see in later exercises.  It is a very common source of mistakes, so keep an eye out for this!_

Notice that, unlike the `while` version, in the `for` version we didn't have to create the `counter` variable ourselves, nor do we have to increment it - this is handled by the `for` loop, which handles the `counter` variable for us.

## Exercise 3: Nesting

In Python, we can place code structures _inside_ each other.  For example, there is nothing to stop us putting a `for` loop inside `for` loop, or an `if` statement inside a `for` loop.

### Loops inside loops

Consider this example:

> ```python
> for i in range(10, 14):
>     print("i is", i)
>     for j in range(20, 24):
>         print("  j is", j)
> ```

There are two `for` loops: an _inner_ and _outer_ loop.  The rules here are the same as always - each loop executes the code in its block multiple times.  This means that for each outer loop value of `i`, the inner loop will iterate over all values of `j`.  So the output will look like this:

> ```text
> i is 1
>   j is 1
>   j is 2
>   j is 3
> i is 2
>   j is 1
>   j is 2
>   j is 3
> i is 3
>   j is 1
>   j is 2
>   j is 3
> ```

Notice that the inner loop's code is indented twice - once to show it is inside the outer loop, and once more to show it is inside the inner loop.

### `if` inside a loop

We can also place an `if` statement inside a loop.  This lets us check a condition at each step and take different actions depending on the value.  For example:

> ```python
> for i in range(1, 6):
>     if i > 3:
>         print(i, "is greater than 3")
>     else:
>         print(i, "is 3 or less")
> ```

```text
1 is 3 or less
2 is 3 or less
3 is 3 or less
4 is greater than 3
5 is greater than 3
```

### `if` inside `if`

Finally, we can place an `if` statement inside another `if` statement.  The inner `if` only runs if the outer condition was true.  This lets us check several conditions one step at a time.  For example:

> ```python
> if age >= 18:
>     if age >= 65:
>         print("Senior")
>     else:
>         print("Adult")
> else:
>     print("Child")
> ```

Here, we only check whether someone is a senior if we have already established that they are an adult.  Note the double indentation on the inner `if` block.
