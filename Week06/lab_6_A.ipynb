## Exercise 1: Object Methods

Last week, we learned about classes and objects, and how they can be used to represent complex data types.

This week, we will learn another important feature: **methods**.  Methods are functions which are defined inside a class, and can be used to access and modify the data stored inside objects.

Here is a simple example of a class with a method.  The class represents a product in a retail system.  It has an attribute which represents the net price (the price without tax), but it also has a method which returns the gross price (with tax).

> ```python
> class Product:
>     def __init__(self, net_price):
>         self.net_price = net_price
>     
>     def gross_price(self):
>         return self.net_price * 1.2
> ```

Here are some key points:
* The `gross_price` method is a function, exactly like the functions we have seen before
* However, it is indented so that it lives _inside_ the `Product` class
* It has an argument called `self`, which refers to the object instance.  We can use this to access the attributes.
* We can use the `self` argument to access the attributes using the dot notation, just like any other object reference

If we want to **call** this method, we use the dot notation that use usually use for the object's attributes, combined with the usual brackets that we apply to function calls.  For example:

> ```python
> product = Produce(100)
> price_with_tax = product.gross_price()
> ```

There is one notable difference about method calls, compared to normal function calls: we do **not** provide a value for the `self` argument.  Python fills this in automatically.  So, method calls always have one fewer argument than the method definition would suggest.  We saw the same thing with the object constructors last week - the `self` argument is ignored in that case too.

### Protected Attributes


One of the goals of object-oriented programming is to keep related data and methods together in one place - that is, inside the class.  The idea is that we can easily see which bits of code might affect an attribute if all that code is kept together.

This can be particularly important to understand the assumptions we can make about a piece of data.  Consider a banking application.  We might have a rule that the balance is never allowed to be negative, and so we might want to write some code which assumes this rule is being followed.  In a large, disorganized program, we might not feel confident that this assumption is true, because _any_ piece of code _anywhere_ in the program could be affecting the balance in ways we weren't expecting.  But if we believe that a strict object-oriented pattern has been used, where we have a `BankAccount` class with a `balance` attribute, we should be able to just look at the methods in the `BankAccount` class to see how the `balance` attribute is treated.

Some stricter object-oriented languages allow you to **enforce** these assumptions, so that **only** the code inside a class is allowed to read or write an attribute.  Python is not strictly object-oriented, but it does have a _convention_ which helps communicate to other developers what assumptions they can make.  The convention is that attributes starting with an underscore (`_`) are **protected** and should **not** be accessed in any way from outside the class code.

Consider this example:

> ```python
> class BankAccount:
>     def __init__(self, balance):
>         if balance >= 0:
>             self._balance = balance
>         else:
>             self._balance = 0
> 
>     def withdraw(self, amount):
>         if amount < self._balance:
>             self._balance -= amount
>     
>     def get_balance(self):
>         return _balance
> ```

There are two things to note about this class:
* It has been written to ensure that `balance` **cannot** be negative, as long as we interact with the object via the methods.
* The `_balance` attribute has an underscore in front of it, which indicates that it should **not** be accessed from outside the class
* We have written a method called `get_balance` to return the balance to outside code if needed, but this method does not allow the balance to be changed

By writing a class this way, we can easily see that the balance will never be negative, and we can make a reasonable assumption that no code outside the class will modify the `_balance` attribute since it is marked as protected.

As a general rule: **make all attributes protected unless they are specifically meant to be modified by other code**.

### Calling Methods Internally and Protected Methods

We have seen how we can call methods on object by using the object instance and the dot notation, for example:

> ```python
> product.set_id("123")
> ```

It is also valid to call methods on an object from _inside_ its class.  In this case, we can use `self` as the object instance, for example:

> ```python
> self.set_id("123")
> ```

This is useful when we have code used in multiple places, and we want to avoid rewriting it.  For example, consider this class, which is the controller for the display on an old CD player.  It displays a line of text, the current track, and the current position in seconds:

```python
class CdDisplay:
    def __init__(self):
        self.display_text = "HELLO"
        self._reset()
    
    def cd_ejected(self):
        self.display_text = "EJECT"
        self._reset()
    
    def stop_pressed(self):
        self.display_text = "STOP"
        self._reset()

    def _reset(self):
        self.track = 1
        self.position_in_seconds = 0
```

The thing to note is the use of the `_reset()` method.  This method resets the track and position to their defaults, which happens in multiple places in the code (including in the constructor).  So, it makes sense to turn this into a method and call it rather than repeat the code.

Also note that the name of the method starts with an underscore.  Just like with attributes, the underscore indicates that the method is _protected_, and it is not supposed to be used by any code external to the class.
