## Exercise 1: Functions

It is often useful to write chunks of code which are _reusable_, so that we can write a block of code once but then run it in many different places throughout the program.  In Python, we can do this using _functions_.

Below is an example of a function being created:

> ```python
> def say_hello():
>     print("Hello, world!")
> ```

This code says: "create a function called `say_hello`.  When the function runs, execute `print("Hello, world!")`".  We will look at this syntax in more detail soon.

If we were to run a program with **only** the code above, what do you think would happen?  The answer is **nothing**.  This code _defines_ a function, but does not _run_ it.  If we want to run the function, we do so by stating the function name, followed by brackets, somewhere in our code:

```python
say_hello()
```

The proper terminology for this is a _function call_.  When we cause the function to run, we say we are _calling_ the function.

_💡 Note: You have called functions before.  `print`, `len` and `range` are all functions.  They are built-in to Python, which is why they didn't need to be defined first._

### Exercise 1.3

Functions become a lot more useful when we can _pass a value_ to a function call.  This means we can reuse a block of code many times, but with different _inputs_.  For example, consider this simple function:

> ```python
> def print_tax(price):
>     print(price * 0.2)
> ```

Let's break this down:
* `def` and the function name (`print_tax`) are just as before.
* The brackets now contains a name, `price`.  This is a _variable name_, and it represents the _input_ to the function.  It tells us that the `price` variable will be available throughout the function's code block.  The proper name for this is an **argument**, or sometimes, a **parameter**.
* There is a code block containing `print`, but we are printing the result of `price` multiplied by 0.2.  Note that, at this stage, **we do not know the value of `price`**.  We are writing the function so that it is a _general purpose, reusable_ function for printing the tax amount, and will work for any price.

Now that we have added an argument (an input parameter), we **must** provide a value for it when we call the function.  For example:

> ```python
> print("Tax for £100 item:")
> print_tax(100)
> 
> print("Tax for £20 item:")
> print_tax(20)
> ```

### Exercise 1.7

So far, all of our functions have printed something.  However, as well as taking input arguments, functions can also **return** an output value.  For example, consider the function we wrote earlier, which calculated and printed tax.  It would be more useful if the function calculated the tax and returned it back to the place it was called as a **value**, instead of printing it.  That way, we could use it in further calculations if we wanted to.

We can do this using Python's `return` keyword.  Consider this code:

> ```python
> def calculate_tax(price):
>     return price * 0.2
> ```

This function _returns_ the calculated tax as a value, so it can be used in further calculations.  The important bit is the `return` keyword, after which is the **value we want to return**.  In this case, that value is a **float**.

_💡 A function **always ends if it reaches a `return` statement**.  If you put more code after the return, it won't have any effect._

We might use the function above like this:

> ```python
> price = 200
> tax = calculate_tax(price)
> total = price + tax
> ```

Notice that the function call to `calculate_tax` is being used as a value, instead of sitting on a line by itself.  In this case, it is being used in a variable assignment, but it could be used like any other value, for example:

> ```python
> total = price + calculate_tax(price)
> ```
