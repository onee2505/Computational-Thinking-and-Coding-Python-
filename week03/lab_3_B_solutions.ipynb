### Exercise 1.1

```python
def login(username, password):
    if username != "mmu":
        print("Incorrect username")
    elif password != "hunter2":
        print("Incorrect password")
    else:
        print("Login successful")

login("admin", "hunter2")
login("mmu", "password")
login("mmu", "hunter2")
```

### Exercise 1.2

```python
def past_tense(word):
    return word + "ed"

print("The past tense of walk is:")
print(past_tense("walk"))
print("The past tense of play is:")
print(past_tense("play"))
print("The past tense of jump is:")
print(past_tense("jump"))
```

### Exercise 1.3

```python
def show_menu():
    print("Choose an option:")
    print("1. Add item")
    print("2. Remove item")
    print("3. Quit")

show_menu()

for i in range(3):
    if i == 0:
        print("You selected Add item")
        show_menu()
    elif i == 2:
        print("You selected Remove item")
        show_menu()
    else:
        print("You selected Quit")
        show_menu()
```

### Exercise 1.4

```python
def find_maximum(temp_list):
    max = -999999999
    for temp in temp_list:
        if temp > max:
            max = temp
    
    return max

monday_temp =  [ 5, 5, 6, 7, 9, 11, 11, 12, 11, 10, 9, 8 ]
tuesday_temp = [ 5, 6, 6, 7, 9, 12, 12, 13, 11, 10, 9, 9 ]
wednesday_temp = [ -1, 0, 0, 2, 3, 4, 5, 5, 4, 4, 3, 3 ]

print("Monday maximum temperature:")
print(find_maximum(monday_temp))
print("Tuesday maximum temperature:")
print(find_maximum(tuesday_temp))
print("Wednesday maximum temperature:")
print(find_maximum(wednesday_temp))

```

### Exercise 1.5

```python
def power(a, b):
    result = 1

    for i in range(b):
        result *= a
    
    return result

print("2 to the power of 6 is:")
print(power(2, 6))
print("3 to the power of 3 is:")
print(power(3, 3))
print("23 to the power of 5 is:")
print(power(23, 5))
```

### Exercise 1.6

```python
def print_dashes(number):
    result = ""

    for i in range(number):
        result += "-"

    print(result)


for i in range(8):
    print_dashes(i + 1)

for i in range(7):
    print_dashes(7 - i)
```
