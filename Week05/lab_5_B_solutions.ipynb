### Exercise 1.1

```python
class DayRecord:
    def __init__(self, day_name, temperatures):
        self.day_name = day_name
        self.temperatures = temperatures

def get_max_temperature(day_record):
    max_temp = -999
    for temp in day_record.temperatures:
        if temp > max_temp:
            max_temp = temp
    return max_temp

monday_temps = DayRecord("Monday", [15, 22, 18, 30, 25])
tuesday_temps = DayRecord("Tuesday", [17, 24, 20, 28, 26])
wednesday_temps = DayRecord("Wednesday", [16, 23, 19, 27, 29])

print("Max temperatures")
print("Monday:", get_max_temperature(monday_temps))
print("Tuesday:", get_max_temperature(tuesday_temps))
print("Wednesday:", get_max_temperature(wednesday_temps))
```

### Exercise 1.2

```python
class EmployeeRecord:
    def __init__(self, employee_name, hourly_rate, hours_worked):
        self.employee_name = employee_name
        self.hourly_rate = hourly_rate
        self.hours_worked = hours_worked

records = [
    EmployeeRecord("Sam", 25, 23),
    EmployeeRecord("Hannah", 28, 20),
    EmployeeRecord("Ali", 28, 32)
]

total_cost = 0

for record in records:
    total_cost += record.hourly_rate * record.hours_worked

print("Total cost:", total_cost)
```

### Exercise 1.3

```python
class BankAccount:
    def __init__(self, balance):
        self.balance = balance


def transaction(source_account, destination_account, amount):
    source_account.balance -= amount
    destination_account.balance += amount

alice_account = BankAccount(100)
bob_account = BankAccount(50)

transaction(alice_account, bob_account, 30)

print("Alice balance:", alice_account.balance)
print("Bob balance:", bob_account.balance)
```

### Exercise 1.4

```python
class Patient:
    def __init__(self, name, age):
        self.name = name
        self.age = age

patients = [
    Patient("Marie Curie", 30),
    Patient("Albert Einstein", 25),
    Patient("Ada Lovelace", 40),
    Patient("Roger Penrose", 31)
]

def check_patient_screening(patient):
    if patient.age >= 30:
        print(patient.name)

print("Patients for screening:")
for patient in patients:
    check_patient_screening(patient)
```

### Exercise 1.5

```python
class UserDirectory:
    def __init__(self, admin_users, basic_users):
        self.admin_users = admin_users
        self.basic_users = basic_users

class User:
    def __init__(self, username, email):
        self.username = username
        self.email = email

user_directory = UserDirectory([
    User("admin", "admin@mmu.ac.uk"),
], [
    User("Jose", "jose@mmu.ac.uk"),
    User("Pat", "pat@mmu.ac.uk")
])

print("*** Admin users ***")
for user in user_directory.admin_users:
    print(user.email)

print("*** Basic users ***")
for user in user_directory.basic_users:
    print(user.email)
```

### Exercise 1.6

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.is_available = True

books = [
    Book("1984", "George Orwell"),
    Book("To Kill a Mockingbird", "Harper Lee"),
    Book("The Great Gatsby", "F. Scott Fitzgerald")
]

def loan_book(books, title):
    found = False

    for book in books:
        if book.title == title:
            found = True
            if book.is_available:
                book.is_available = False
                print("Loan started: ", title)
            else:
                print("Book is not available: ", title)
    
    if found == False:
        print("Book not found: ", title)

loan_book(books, "1984")
loan_book(books, "1984")
loan_book(books, "1983")
    
```
