## Exercise 1: Classes and Objects

Classes and objects are used to create **complex data types**.  These are data types which have **multiple parts**.  For example, if we want to represent a *person* in our system, we may want to store a first name, a last name, a date of birth etc. as part of one single data item.  These parts are called _attributes_.

To define a complex data type like this, we use **classes**.  A class is a built-in feature of Python which allows us to describe a complex data type.  Look at the example below, which represents a player's state in a video game.

> ```python
> class GameState:
>     def __init__(self):
>         self.score = 0
>         self.level = 1
>         self.lives = 3
> ```

We discuss this syntax in more detail soon, but here is an overview of what each line means:
* `class GameState`: We are creating a new type called `GameState`
*  `def __init__(self):`: The definition of a function called `__init__`, which has an argument called `self`
*  `self.score`: The type will have an attribute called `score`, and it starts at zero
*  `self.level`: The type will have an attribute called `level`, and it starts at one
*  `self.lives`: The type will have an attribute called `score`, and it starts at three

By itself, this code doesn't cause any effect.  It simple defines the class.  If we want to _use_ the class, we must **instantiate** it into a **object**.  Objects are the actual data items which are creating from classes, and they hold the attribute values.

To create an object instance from this class, we use this syntax:

> ```python
> game_state = GameState()
> ```

This code stores a `GameState` object instance in the `game_state` variable.  Notice that, essentially, we are using the class name like it is a function: we write the class name, followed by brackets.  For now, the brackets are empty.

Now, we can actually use the data stored in the object.  For example, we might want to print the value of the `score` attribute on this object.  To access the attributes, we use the dot (`.`) like this:

> ```python
> print(game_state.score)
> ```

So the format is: `object.attribute`.

In this exercise, we will start to define a class ourselves.

Looking at the class from the previous exercise again, let's overview the syntax in a bit more detail:

> ```python
> class RoomSettings:
>     def __init__(self):
>         self.temperature = 18
>         self.lighting_level = 0
> ```

* Classes are defined using the `class` keyword, followed by the name of the class.  This follows the usual naming rules, so spaces are not allowed.  Conventionally, class names start with capitals and also use capital letters to separate words, rather than using underscores like variable names.
* The line ends with a colon to indicate that everything indented below it "belongs" to the class.
* Inside the class (and therefore, indented) we MUST define a function called `__init__`.  Note there are **two** underscores at the start and **two** at the end.  This is a special name which indicates that this function is the _constructor_ for the class, and will get called whenever new objects are instantiated.
* The `__init__` function MUST have an argument called `self`.  When an object is created from the class, `self` will refer to the new object instance in memory.
* In the body of the constructor function we can do whatever we want to initialize the new object.  Usually, this involves setting attribute values - otherwise, the object won't have any attributes!
* Since the object being created is always brand new, it won't have any attributes at first, so every time we write a line like `self.temperature = 18`, we are _creating_ the temperature attribute on the object, as well as setting its initial value.

### Exercise 1.5

So far we have seen classes which create attributes that have a default value, and we've also seen how we can write code to update an attribute value after an object is created.  A more common pattern to have a class which allows **arguments to be passed into its constructor function**, which are then used to initialize the attributes.

For example, consider this class:

> ```python
> class User:
>     def __init__(self, first_name, last_name):
>         self.first_name = first_name
>         self.last_name = last_name
>         self.login_attempts = 0
> ```

Note that unlike previous examples, the `__init__` function now has **three** arguments.  The first **must** be `self`, but the next two are the values for `first_name` and `last_name` which we want the object to have when it is created.  It makes sense to write it like this, because for a human user there is unlikely to be a suitable "default" name which we could start with - instead, we will _always_ want to set the name to something specific, so we will **require** the code which is instantiating the object to specify a name.

Also note the slightly confusing line `self.first_name = first_name`.  This mentions `first_name` twice, but is technically referring to two different things.  `self.first_name` is the attribute on the object which we are setting.  `first_name` alone refers to the input argument.  So this line says, "Create an attribute on the object called `first_name`, and set it to have whatever value was passed into the `first_name` argument of the function".

Finally, notice that we don't have to initialize _all_ of the attributes using arguments.  There is a third attribute called "login_attempts" which is still initialized to zero, because all users will start with zero login attempts.  So there is no need to require a value to be passed to the constructor.

When we write a constructor like this, we MUST provide values for the arguments when we create object instances.  For example:

> ```python
> user1 = User("William", "Shakespeare")
> ```

Note that we **ignore** the `self` argument to the constructor (Python deals with this automatically).  We just pass in the two arguments specified after `self` in the constructor, which are `first_name` and `last_name`.
