## Exercise 1


In this exercise you will write a program which finds the maximum temperature for a particular day.  There is a class which represents several temperatures for a specific day.

* Create a class called `DayRecord`, with string attribute `day_name`, and list of integers `temperatures`.
* Instantiate three `DayRecord` objects.  The three objects represent the temperatures for Monday, Tuesday and Wednesday, and the data you should instantiate them with is:

> ```text
> Monday: 15, 22, 18, 30, 25
> Tuesday: 17, 24, 20, 28, 26
> Wednesday: 16, 23, 19, 27, 29
> ```

* Write a function which takes a `DayRecord` object as an argument, finds the maximum temperature in the `temperatures` attribute, and returns it.
* Print the maximum temperature for each of the three days.

You output should look something like this:

> ```text
> Max temperatures
> Monday: 30
> Tuesday: 28
> Wednesday: 29
> ```

### Exercise 1.6

In this exercise you will simulate a library system, in which we can take a book out on loan, after which the book cannot be loaned again.

**Write a program which does the following:**
* Write a class called `Book` with the attributes `title`, `author` and `is_available`.  `is_available` defaults to `True` for every new object.
* Create a list of books with the following data:

| Title                      | Author              |
|----------------------------|---------------------|
| 1984                       | George Orwell       |
| To Kill a Mockingbird      | Harper Lee          |
| The Great Gatsby           | F. Scott Fitzgerald |

* Write a function called `loan_book` which takes a list of `Book` objects and a title as arguments
  * The function should search through the list for the book with the correct title
  * If found and the book is available, set the book's `is_available` to `False` and print "Loan started".
  * If found and the book is not available, print "Book is not available.".
  * If the book title is not found, print "Book not found".
* Call `loan_book` three times with the title "1984", then "1984" again, then "1983".

The output should look something like this:

> ```text
> Loan started:  1984
> Book is not available:  1984
> Book not found:  1983
> ```
