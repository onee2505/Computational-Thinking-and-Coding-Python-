### Exercise 1.7

```python
class OrderItem:
    def __init__(self, product_name, quantity, unit_price):
        self.product_name = product_name
        self.quantity = quantity
        self.unit_price = unit_price
        self.total_price = quantity * unit_price

item = OrderItem("Bananas", 8, 15)

# The code below will print some output once you have written your class and created the `item` variable.

print("Item name:", item.product_name)
print("Order total price (p):", item.total_price)
```

### Exercise 1.8

```python
class Movie:
    def __init__(self, title, rating):
        self.title = title
        self.rating = rating

def print_movie(movie):
    print("=== Movie ===")
    print("Title:", movie.title)
    print("Rating:", movie.rating)

movies = [
    Movie("Inception", 8.8),
    Movie("The Godfather", 9.2),
    Movie("Citizen Kane", 8.3)
]

for movie in movies:
    print_movie(movie)
```
